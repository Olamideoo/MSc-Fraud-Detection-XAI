```python
# Data manipulation
import pandas as pd
import numpy as np

# Data preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
# Model Training

#This notebook trains and evaluates multiple machine learning models for fraud detection using the engineered dataset. 
#The workflow includes data preparation, experiment tracking with MLflow, model training, performance evaluation, and model selection. The best-performing model will be used in the explainability stage of the proposed Explainable AI (XAI) fraud detection framework.
# Logistic Regression
# Random Forest
# XGBoost
# Model Evaluation
%pip install lightgbm
# Data manipulation
import pandas as pd
import numpy as np

# Data preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Machine learning models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# Model evaluation
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Experiment tracking
import mlflow
import mlflow.sklearn

# Model persistence
import joblib

# Timing
import time

# Visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
# Loading Datase
# Load the engineered dataset
data = pd.read_csv("../Data/engineered_fraud_dataset.csv")

# Display the first five rows
data.head()
## Dataset Overview

#The engineered dataset is loaded and inspected to confirm that all features have been successfully imported and are ready for model training. This step verifies the dataset structure, identifies any missing values, and confirms the data types of each variable.
# Display the first five rows
data.head()

# Display dataset dimensions
print("Dataset Shape:", data.shape)

# Display column names
print("Columns:")
print(data.columns.tolist())

# Check for missing values
print("Missing Values:")
print(data.isnull().sum())

# Check data types
print("Data Types:")
print(data.dtypes)
## Identify Feature Types

#The dataset contains both numerical and categorical variables. Identifying feature types is necessary before applying preprocessing techniques such as one-hot encoding and feature scaling.
# Separate numerical and categorical columns
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = data.select_dtypes(include=['object']).columns.tolist()

print("Numerical Features:")
print(numerical_cols)

print("\nCategorical Features:")
print(categorical_cols)
## Encode Categorical Variables

#Machine learning algorithms require numerical input. Therefore, the categorical variables are transformed into numerical representations using one-hot encoding.
# One-hot encode categorical variables
data = pd.get_dummies(
    data,
    columns=categorical_cols,
    drop_first=True
)

print("Dataset Shape After Encoding:", data.shape)

data.head()
bool_columns = data.select_dtypes(include='bool').columns

data[bool_columns] = data[bool_columns].astype(int)

print(data.head())
## Define Features and Target Variable

#The dataset is divided into input features (X) and the target variable (y). The target variable is `fraud_flag`, while all remaining variables are used as predictors for fraud detection.
# Define features and target variable
X = data.drop('fraud_flag', axis=1)
y = data['fraud_flag']

print("Feature Matrix Shape:", X.shape)
print("Target Variable Shape:", y.shape)
#checking the distribution of the target variable

print("Target Variable Distribution:")
print(y.value_counts())

print("\nTarget Variable Percentage:")
print(y.value_counts(normalize=True) * 100)
## Train-Test Split

#The dataset is divided into training and testing subsets using an 80:20 ratio. Stratified sampling is applied to preserve the class distribution of fraudulent and non-fraudulent transactions in both datasets.
from sklearn.model_selection import train_test_split

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y)


# Display dataset sizes
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
# Model Training and Evaluation

## 1. Baseline Model Evaluation

### Logistic Regression (Baseline)
#- Train model
#- Evaluate model
#- Record results

### LightGBM (Default Parameters)
#- Train model
#- Evaluate model
#- Record results

### Baseline Model Comparison
#(results_df)
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit only on the training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same transformation to the test data
X_test_scaled = scaler.transform(X_test)
### LOGISTIC REGRESSION MODEL
# Create an empty list to store model results
results = []
import time

# Start training timer
start_train = time.time()

# Create the model
lr_model = LogisticRegression(
    random_state=42,
    max_iter=1000
)

# Train the model
lr_model.fit(X_train_scaled, y_train)

# End training timer
end_train = time.time()

# Calculate training time
training_time = end_train - start_train

print("Logistic Regression trained successfully!")
print(f"Training Time: {training_time:.4f} seconds")
## prediction and evaluation 

# Start prediction timer
start_pred = time.time()

# Make predictions
y_pred_lr = lr_model.predict(X_test_scaled)
y_prob_lr = lr_model.predict_proba(X_test_scaled)[:, 1]

# End prediction timer
end_pred = time.time()

# Calculate prediction time
prediction_time = end_pred - start_pred

print(f"Prediction Time: {prediction_time:.6f} seconds")
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred_lr)
precision = precision_score(y_test, y_pred_lr)
recall = recall_score(y_test, y_pred_lr)
f1 = f1_score(y_test, y_pred_lr)
roc_auc = roc_auc_score(y_test, y_prob_lr)

# Display the results
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")
print(classification_report(y_test, y_pred_lr))
cm = confusion_matrix(y_test, y_pred_lr)
print(cm)
results.append({
    "Model": "Logistic Regression",
    "Training Time (s)": training_time,
    "Prediction Time (s)": prediction_time,
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1,
    "ROC-AUC": roc_auc
})
## XGBOOST MODEL
import time

# Start training timer
start_train = time.time()

# Create the XGBoost model
xgb_model = XGBClassifier(
    random_state=42,
    eval_metric='logloss'
)

# Train the model
xgb_model.fit(X_train, y_train)

# End training timer
end_train = time.time()

# Calculate training time
training_time = end_train - start_train

print("XGBoost model trained successfully!")
print(f"Training Time: {training_time:.4f} seconds")
# Start prediction timer
start_pred = time.time()

# Make predictions
y_pred_xgb = xgb_model.predict(X_test)
y_prob_xgb = xgb_model.predict_proba(X_test)[:, 1]

# End prediction timer
end_pred = time.time()

# Calculate prediction time
prediction_time = end_pred - start_pred

print(f"Prediction Time: {prediction_time:.6f} seconds")
##Evaluation metrics

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred_xgb)
precision = precision_score(y_test, y_pred_xgb)
recall = recall_score(y_test, y_pred_xgb)
f1 = f1_score(y_test, y_pred_xgb)
roc_auc = roc_auc_score(y_test, y_prob_xgb)

# Display the results
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")

##### classisfication report 
print(classification_report(y_test, y_pred_xgb))
#### confusion matrix
cm = confusion_matrix(y_test, y_pred_xgb)
print(cm)
#### Store Result
results.append({
    "Model": "XGBoost",
    "Training Time (s)": training_time,
    "Prediction Time (s)": prediction_time,
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1,
    "ROC-AUC": roc_auc})
## lightGBM Model
import time

# Start training timer
start_train = time.time()

# Create the LightGBM model
lgbm_model = LGBMClassifier(
    random_state=42
)

# Train the model
lgbm_model.fit(X_train, y_train)

# End training timer
end_train = time.time()

# Calculate training time
training_time = end_train - start_train

print("LightGBM model trained successfully!")
print(f"Training Time: {training_time:.4f} seconds")
# Start prediction timer
start_pred = time.time()

# Make predictions
y_pred_lgbm = lgbm_model.predict(X_test)
y_prob_lgbm = lgbm_model.predict_proba(X_test)[:, 1]

# End prediction timer
end_pred = time.time()

# Calculate prediction time
prediction_time = end_pred - start_pred

print(f"Prediction Time: {prediction_time:.6f} seconds")
accuracy = accuracy_score(y_test, y_pred_lgbm)
precision = precision_score(y_test, y_pred_lgbm)
recall = recall_score(y_test, y_pred_lgbm)
f1 = f1_score(y_test, y_pred_lgbm)
roc_auc = roc_auc_score(y_test, y_prob_lgbm)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")
#### Classification 
print(classification_report(y_test, y_pred_lgbm))
### confusion matrix
cm = confusion_matrix(y_test, y_pred_lgbm)
print(cm)
### Store the data
results.append({
    "Model": "LightGBM",
    "Training Time (s)": training_time,
    "Prediction Time (s)": prediction_time,
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1,
    "ROC-AUC": roc_auc
})
## Comparing the models 
results_df = pd.DataFrame(results)
results_df
## Model Optimization Using Hyperparameter Tuning

## Hyperparameter Tuning - XGBoost

### Hyperparameter Tuning - LightGBM

### Optimized Model Comparison

from sklearn.model_selection import RandomizedSearchCV
# Define the hyperparameter search space
xgb_param_grid = {
    "n_estimators": [100, 200, 300],
    "max_depth": [3, 5, 7],
    "learning_rate": [0.01, 0.05, 0.1],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]}
### RandomizedSearchCV object
# Create the Randomized Search object
xgb_random = RandomizedSearchCV(
    estimator=XGBClassifier(
        random_state=42,
        eval_metric="logloss"
    ),
    param_distributions=xgb_param_grid,
    n_iter=20,
    scoring="f1",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
### ESTIMATOR
estimator=XGBClassifier(...)
param_distributions=xgb_param_grid
n_iter=20
### performing hyperparameter tuning 
# Perform hyperparameter tuning
xgb_random.fit(X_train, y_train)

print("Hyperparameter tuning completed!")
print("Best Parameters:")
print(xgb_random.best_params_)
print("Best F1 Score:")
print(xgb_random.best_score_)
### Train the Optimized XGBoost Model
import time
from xgboost import XGBClassifier

# Start training timer
start_train = time.time()

# Create the optimized model
xgb_optimized = XGBClassifier(
    n_estimators=100,
    max_depth=5,
    learning_rate=0.05,
    subsample=1.0,
    colsample_bytree=1.0,
    random_state=42,
    eval_metric="logloss"
)

# Train the model
xgb_optimized.fit(X_train, y_train)

# End training timer
end_train = time.time()

training_time = end_train - start_train

print("Optimized XGBoost trained successfully!")
print(f"Training Time: {training_time:.4f} seconds")
### PREDICTION
# Start prediction timer
start_pred = time.time()

# Make predictions
y_pred_xgb_opt = xgb_optimized.predict(X_test)
y_prob_xgb_opt = xgb_optimized.predict_proba(X_test)[:, 1]

# End prediction timer
end_pred = time.time()

# Calculate prediction time
prediction_time = end_pred - start_pred

print(f"Prediction Time: {prediction_time:.6f} seconds")
# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred_xgb_opt)
precision = precision_score(y_test, y_pred_xgb_opt)
recall = recall_score(y_test, y_pred_xgb_opt)
f1 = f1_score(y_test, y_pred_xgb_opt)
roc_auc = roc_auc_score(y_test, y_prob_xgb_opt)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")
print(classification_report(y_test, y_pred_xgb_opt))
cm = confusion_matrix(y_test, y_pred_xgb_opt)
print(cm)
optimized_results = []
optimized_results.append({
    "Model": "Optimized XGBoost",
    "Training Time (s)": training_time,
    "Prediction Time (s)": prediction_time,
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1,
    "ROC-AUC": roc_auc
})
### Hyperparameter Tuning – LightGBM
# Define the hyperparameter search space
lgbm_param_grid = {
    "n_estimators": [100, 200, 300],
    "learning_rate": [0.01, 0.05, 0.1],
    "max_depth": [3, 5, 7],
    "num_leaves": [31, 50, 70],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0]
}
### RandomizedSearchCV Object
# Create the Randomized Search object
lgbm_random = RandomizedSearchCV(
    estimator=LGBMClassifier(
        random_state=42
    ),
    param_distributions=lgbm_param_grid,
    n_iter=20,
    scoring="f1",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
### Performing hyperparameter tuning 
# Perform hyperparameter tuning
lgbm_random.fit(X_train, y_train)

print("LightGBM hyperparameter tuning completed!")
print("Best Parameters:")
print(lgbm_random.best_params_)
print("Best F1 Score:")
print(lgbm_random.best_score_)
### OPTIMIZING THE MODEL 
import time
from lightgbm import LGBMClassifier

# Start training timer
start_train = time.time()

# Create the optimized LightGBM model
lgbm_optimized = LGBMClassifier(
    n_estimators=300,
    learning_rate=0.01,
    max_depth=5,
    num_leaves=70,
    subsample=0.8,
    colsample_bytree=1.0,
    random_state=42
)

# Train the model
lgbm_optimized.fit(X_train, y_train)

# End training timer
end_train = time.time()

training_time = end_train - start_train

print("Optimized LightGBM trained successfully!")
print(f"Training Time: {training_time:.4f} seconds")
# Start prediction timer
start_pred = time.time()

y_pred_lgbm_opt = lgbm_optimized.predict(X_test)
y_prob_lgbm_opt = lgbm_optimized.predict_proba(X_test)[:, 1]

end_pred = time.time()

prediction_time = end_pred - start_pred

print(f"Prediction Time: {prediction_time:.6f} seconds")
accuracy = accuracy_score(y_test, y_pred_lgbm_opt)
precision = precision_score(y_test, y_pred_lgbm_opt)
recall = recall_score(y_test, y_pred_lgbm_opt)
f1 = f1_score(y_test, y_pred_lgbm_opt)
roc_auc = roc_auc_score(y_test, y_prob_lgbm_opt)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")
print(classification_report(y_test, y_pred_lgbm_opt))
cm = confusion_matrix(y_test, y_pred_lgbm_opt)
print(cm)
optimized_results.append({
    "Model": "Optimized LightGBM",
    "Training Time (s)": training_time,
    "Prediction Time (s)": prediction_time,
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1,
    "ROC-AUC": roc_auc})
## Final Model Comparison and Selection
import pandas as pd

# Baseline model results
results_df = pd.DataFrame(results)

# Optimized model results
optimized_results_df = pd.DataFrame(optimized_results)
print("Baseline Model Performance")
display(results_df)

print("Optimized Model Performance")
display(optimized_results_df)
print("Baseline Model Performance")
display(results_df)

print("Optimized Model Performance")
display(optimized_results_df)
# Combine baseline and optimized model results
final_results = pd.concat(
    [results_df, optimized_results_df],
    ignore_index=True
)

# Display the complete comparison table
print("Final Model Comparison")
display(final_results)
# Select the final model
final_model = lgbm_optimized

print("Final Selected Model: Optimized LightGBM")
import joblib

joblib.dump(lgbm_optimized, "final_lightgbm_model.joblib")
#grid_search = GridSearchCV(...)
### Model Selection

#The baseline and optimized machine learning models were compared using Accuracy, Precision, Recall, F1 Score, ROC-AUC, training time, and prediction time.

#Hyperparameter tuning improved the performance of both XGBoost and LightGBM compared to their default configurations.

#The Optimized LightGBM model achieved the strongest overall balance between classification performance and computational efficiency, recording the highest Accuracy (95.30%), Precision (82.77%), and F1 Score (80.74%), while maintaining a competitive Recall (78.80%) and ROC-AUC (97.79%).

#Based on these results, the Optimized LightGBM model was selected as the final model for the explainable AI fraud detection framework. This model will be used in the subsequent explainability, experiment tracking, and deployment stages of the project.
# Machine learning models
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# Model evaluation
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Experiment tracking
import mlflow
import mlflow.sklearn

# Model persistence
import joblib

# Timing
import time

# Visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
```
